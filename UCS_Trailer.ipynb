```python
# Importing required libraries
import heapq
import networkx as nx
import matplotlib.pyplot as plt
```

```python
# definition of the search space (graph), initial state (start)
# and goal state (goal)
def uniform_cost_search(graph, start, goal):

  # priority queue: (cost, node, path)
  frontier = [(0, start, [start])]

  # keeps track of all nodes that have already been fully expanded
  explored = set()


  # let the algorithm keep searching as long as there are viable paths left
  # in the frontier
  while frontier:
    cost, node, path = heapq.heappop(frontier)

    # skip nodes already expanded
    if node in explored:
      continue
    explored.add(node)

    # Goal check when node is expanded
    # Return the cost and path when goal is reached
    if node == goal:
      return cost, path

    # examine each neighbour and the cost of reaching it
    for neighbor, edge_cost in graph.get(node, []):

      if neighbor not in explored:
        # calculate the new total cost and extend the path
        new_cost = cost + edge_cost
        new_path = path + [neighbor]

        #add the route to the priority queue
        heapq.heappush(frontier, (new_cost, neighbor, new_path))

  return float ('inf'), [] # if no path found, return infinity

# Road network with neighbouring town and distance
graph = {
    'Nairobi': [('Nakuru', 161), ('Maai Mahiu', 53.8)],
    'Nakuru' : [('Kericho', 108), ('Molo', 48.5), ('Eldoret', 158), ('Nyahururu', 63.5)],
    'Kericho': [('Kapsoit', 11.3), ('Ahero', 60.9)],
    'Eldoret': [('Kapsabet', 47), ('Webuye', 72)],
    'Nyahururu': [('Molo', 111)],
    'Maai Mahiu':[('Narok', 89)],
    'Molo': [('Kericho', 64.2)],
    'Kapsoit': [('Kisumu', 72.9), ('Ahero', 49.9)],
    'Ahero': [('Kisumu', 23.5)],
    'Kapsabet': [('Kisumu', 80.3)],
    'Webuye': [('Kakamega', 48.5)],
    'Bomet': [('Kisii', 84), ('Litein', 46)],
    'Kakamega': [('Kisumu', 51)],
    'Kisii': [('Oyugis', 28)],
    'Litein': [('Kericho', 33)],
    'Narok': [('Bomet', 81)],
    'Oyugis': [('Ringa', 14)],
    'Ringa': [('Sondu', 22)],
    'Sondu': [('Ahero', 29.5)],
    'Kisumu': []
}

# Search for the lowest-cost route from NBI-KSM
cost, path = uniform_cost_search(graph, 'Nairobi', 'Kisumu')

#Display output
print(f"Optimal cost: {cost: .1f}")
print(f"Optimal path: {' --> '.join(path)}")
```
